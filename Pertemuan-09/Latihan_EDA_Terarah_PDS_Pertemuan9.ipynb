## Bagian 5 — Ringkasan Temuan EDA

```python
# Ringkasan (JANGAN diubah)
ringkasan = pd.DataFrame(catatan, columns=["jenis", "isi"])
print(ringkasan["jenis"].value_counts().reindex(["temuan", "dugaan", "keterbatasan"]).fillna(0).astype(int).to_string())
for jenis in ["temuan", "dugaan", "keterbatasan"]:
    print(f"\n=== {jenis.upper()} ===")
    for i, isi in enumerate(ringkasan.loc[ringkasan["jenis"] == jenis, "isi"], 1):
        print(f"{i}. {isi}")
```
