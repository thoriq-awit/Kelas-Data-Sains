## Bagian 3 — Perbaiki Grafik yang Menyesatkan
Sel berikut menghasilkan grafik yang **menyesatkan**. Jalankan, sebutkan minimal **tiga** masalahnya, lalu buat versi yang jujur.

```python
# Grafik menyesatkan (JANGAN diubah - buat perbaikannya di sel berikutnya)
rata = df.groupby("perangkat_utama")["tingkat_ketertarikan"].mean()
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(rata.index, rata.values, color=["red", "green", "blue"])
ax.set_ylim(2.9, 4.1)
ax.set_title("Pengguna Laptop Jauh Lebih Tertarik Data Sains!")
plt.show()
print(df["perangkat_utama"].value_counts().to_string())
```
